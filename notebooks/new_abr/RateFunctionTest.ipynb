```python
import numpy as np
from reactionmodel.model import Model, Reaction, Species
from reactionmodel.propensity import construct_propensity_function
```

```python
A = Species('A')
B = Species('B')
C = Species('C')
D = Species('D')

r1 = Reaction([A, B], [C], k=1.0)
r2 = Reaction([(C, 2)], [D], k=0.5)

m = Model([A,B,C,D], [r1, r2])
```

```python
propensity_function = m.get_propensities_function(jit=True)
```

```python
propensity_function = construct_propensity_function(m.get_k(jit=True), m.kinetic_order(), jit=False)
```

```python
propensity_function(0.0, m.make_initial_condition({'A': 10, 'B': 3, 'C': 2, 'D': 0}))
```

```python
from numba import jit

@jit(nopython=True)
def kwarg_test(a, b=1, c=2):
    """
    Example function to test numba's handling of kwargs.
    """
    return a + b
```

```python
kwarg_test(1.0, c=None)
```

```python
@jit(nopython=True)
def state_dependent_k(t, y):
    return 0.5 * np.prod(y[:2])

r3 =  Reaction([(C, 2)], [D], k=state_dependent_k)
```

```python
m2 = Model([A,B,C,D], [r1, r3])
propensity_function = m2.get_propensities_function(jit=True)
propensity_function(0.0, m2.make_initial_condition({'A': 10, 'B': 3, 'C': 2, 'D': 0}))
```
