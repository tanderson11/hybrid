```python
import numpy as np
import matplotlib.pyplot as plt
from hybrid.eulermaruyama import em_solve_ivp
```

```python
sigma = 1.0
mu = 10.0
tau = 0.05

def mu_sigma_f(t, y, dt, dW, rng):
    return (mu-y)/tau, np.array([dW(sigma * np.sqrt(dt), rng)])
```

```python
rng = np.random.default_rng()
res = em_solve_ivp(mu_sigma_f, [0.0, 1000.0], np.array([0.0]), rng, dt=0.0001)
```

```python
fig, ax = plt.subplots(1, 1, figsize=(8, 4))
ax.plot(res.t, res.y, lw=2)
```
